```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("L'environnent est prêt")
```

```python
# Création d'un petit dataset fictif
data = {
    'nom': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age': [25, 30, 35, 28, 22],
    'salaire': [3000, 4500, 5200, 3800, 2900]
}

df = pd.DataFrame(data = data)
df
```

```python
df.describe()

plt.figure(figsize=(5, 5))
sns.barplot(x='nom', y='salaire', data=df)
plt.title('Salaire par personne')
plt.xlabel('Nom')
plt.ylabel('Salaire (€)')
plt.show()
```
